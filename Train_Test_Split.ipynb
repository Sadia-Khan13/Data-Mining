Train test split and feature selection

```python
import pandas as pd
```

```python
data = pd.read_csv("Diabetes Missing Data.csv")
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
data["Glucose"].fillna(data["Glucose"].mean(), inplace=True)
data["Diastolic_BP"].fillna(data["Diastolic_BP"].mean(), inplace=True)
data["Skin_Fold"].fillna(data["Skin_Fold"].mean(), inplace=True)
data["Serum_Insulin"].fillna(data["Serum_Insulin"].mean(), inplace=True)
data["BMI"].fillna(data["BMI"].mean(), inplace=True)
```

```python
input_variable = data.iloc[: , :-1]
output_variable= data["Class"]
```

```python
input_variable.shape , output_variable.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(input_variable , output_variable , test_size=0.25)
```

Feature selection

```python
from mlxtend.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
import numpy as np
```

```python
fs = SequentialFeatureSelector(lr , k_features =8  , forward=False)
fs.fit(x_train,y_train)
```

```python
fs.feature_names
```

```python
fs.k_feature_names_ #selected features name
```

```python
fs.k_score_ #to check accuracy
```
