label Encoding convert (nominal data into numeric data)

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.DataFrame({"Name":["sadia" , "cow" , "buttderfly" , "cat" , "ball"]})
df.head()
```

```python
le = LabelEncoder()
df["Name"] = le.fit_transform(df["Name"])
```
