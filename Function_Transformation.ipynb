Function transformation
(use when outliers have most imp data that can not be droped so we use this method)

```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
from sklearn.preprocessing import FunctionTransformer
```

```python
data = pd.read_csv("Dataset Salary 2024.csv")
data.head() 
```

```python
data.info()
```

```python
ft = FunctionTransformer(func=np.log1p)
```

```python
ft.fit(data["salary_in_usd"])
```

```python
data["salary_in_usd_ft"]=ft.transform(data["salary_in_usd"])
```

```python
data.head()
```

```python
plt.figure(figsize=(15 , 5))
plt.subplot(1 , 2 , 1)
plt.title("Before")
sns.distplot(data["salary_in_usd"])

plt.subplot(1 , 2 , 2)
plt.title("After")
sns.distplot(data["salary_in_usd_ft"])

plt.show()
```
