REMOVING OUTLIERS USING IQR METHOD

GENERAL FORM OF IQR METHOD
Q1=25TH PERCENTILE
Q3=75TH PERCENTILE
IQR = Q3-Q1
MIN = Q1-(1.5*IQR)
MAX = Q3+(1.5*IQR)


```python
data.shape
```

```python
Q1 = data["salary_in_usd"].quantile(0.25)
Q3 = data["salary_in_usd"].quantile(0.75)
```

```python
IQR = Q3-Q1
MIN = Q1-(1.5*IQR)
MAX = Q3+(1.5*IQR)
MIN , MAX
```

```python
df = data[data["salary_in_usd"]<= MAX]
```

```python
df.head()
```

```python
df.shape
```

```python
sns.boxplot(df["salary_in_usd"])
plt.show()
```

```python
sns.distplot(df["salary_in_usd"]) 
plt.show()
```

```python
sns.boxplot(data["work_year"])
plt.show()
```

```python
#removing outliers in work years

q1 = data["work_year"].quantile(0.25)
q3 = data["work_year"].quantile(0.75)
iqr = q3-q1
mini = q1 - (1.5*iqr)
maxi = q3 + (1.5*iqr)
mini , maxi
```

```python
data = data[data["work_year"] >= mini]
```

```python
sns.boxplot(data["work_year"])
plt.show()
```

REMOVING OUTLIERS USING SIMPLE METHOD

```python
df = pd.read_csv("Dataset Salary 2024.csv")
df.head(5)
```

```python
df.describe()
```

```python
sns.boxplot(df["salary_in_usd"])
plt.show()
```

```python
mini_value = df["salary_in_usd"].mean() - (3*df["salary_in_usd"].std())
max_value = df["salary_in_usd"].mean() + (3*df["salary_in_usd"].std())
mini_value , max_value
```

```python
df = df[df["salary_in_usd"]<=max_value]
```

```python
sns.boxplot(df["salary_in_usd"])
plt.show()
```

USING Z SCORE METHOD 

```python
z_score = (df["work_year"] - df["work_year"].mean())/(df["work_year"].std())
z_score
```
