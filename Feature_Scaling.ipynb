STANDARDIZATION METHOD

```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("Dataset Salary 2024.csv")
data.head()
```

```python
data.describe()
```

```python
ss =StandardScaler()
ss.fit(data[["salary_in_usd"]])
```

```python
data["salary_in_usd_ss"]=ss.transform(data[["salary_in_usd"]])
```

```python
data.head()
```

```python
plt.figure(figsize = (15 , 6))
plt.subplot(1,2,1)
plt.title("Before")
sns.distplot(data["salary_in_usd"])

plt.subplot(1,2,2)
plt.title("After")
sns.distplot(data["salary_in_usd_ss"])

plt.show()
```

In the above graph values are compressed(rescaled) but the nature of data remain same

NORMALIZATION

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
mm = MinMaxScaler()
mm.fit(data[["remote_ratio"]])
```

```python
data["remote_ratio_mm"] = mm.transform(data[["remote_ratio"]])
```

```python
data.head()
```

```python
plt.figure(figsize = (15 , 6))
plt.subplot(1,2,1)
plt.title("Before")
sns.distplot(data["remote_ratio"])

plt.subplot(1,2,2)
plt.title("After")
sns.distplot(data["remote_ratio_mm"])

plt.show()
```

In the above graph values are compressed between 0 and 1
