One Hot Encoding using get dummies

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
data=pd.read_csv("index.csv")
data.head(5)
```

```python
data.isnull().sum()
```

```python
data["card"].fillna(data["card"].mode()[0] , inplace=True)
```

```python
data.isnull().sum()
```

```python
en = data["cash_type"]
df=pd.get_dummies(en)
df
```

```python
df.head()
```

```python
data =pd.concat([df , data] , axis = "columns")
```

```python
data.head()
```

```python
data.drop("cash_type" , axis = "columns" , inplace = True)
```
